# Dataset exploration
Counts boxes per class and inspects box-size distribution of the YOLO labels in `data/labels`. Run from the repository root.

```python
from pathlib import Path
import pandas as pd

rows = []
for split in ('train', 'val', 'test'):
    for f in Path('data/labels', split).glob('*.txt'):
        for line in f.read_text().splitlines():
            parts = line.split()
            if len(parts) == 5:
                c, x, y, w, h = int(parts[0]), *map(float, parts[1:])
                rows.append(dict(split=split, image=f.stem, cls=c, w=w, h=h))
df = pd.DataFrame(rows)
print(df.groupby(['split', 'cls']).size() if len(df) else 'No labels found yet')
```

```python
if len(df):
    print(df[['w', 'h']].describe())
    print('boxes per image:\n', df.groupby(['split', 'image']).size().groupby('split').describe())
```

Check: (1) every split has images from *different* trees/sessions (no leakage); (2) small normalised box sizes suggest raising `model.imgsz` or enabling tiled inference.
